3. np.mean() vs np.nanmean()
Normal mean:

```python
scores = np.array([150, 180, np.nan, 200, 170])

print(np.mean(scores))
```

```python
#NaN ignore karke mean:
print(np.nanmean(scores))
```

Function	Work
np.mean()	Normal average
np.nanmean()	NaN values ignore karke average
 150+180+200+170/4=175

np.sum() vs np.nansum()

```python
scores = np.array([150, 180, np.nan, 200, 170])

print(np.sum(scores))
print(np.nansum(scores))
```

np.nansum() missing values ko ignore karke remaining values ka sum calculate karta hai.

```python
#np.nan_to_num() – Replace missing values
scores = np.array([150, 180, np.nan, 200, 170])

clean_scores = np.nan_to_num(scores)

print(clean_scores)
```

By default, NaN ko zero se replace karta hai.


```python
#Custom replacement:
clean_scores = np.nan_to_num(scores, nan=100)

print(clean_scores)
```

Important business concept: Har missing value ko zero se replace nahi karna chahiye. Missing score ka matlab zero score nahi hota.

3. Random seed – Reproducibility
Industry mein same experiment ko repeat karna important hota hai.

```python
rng1 = np.random.default_rng(42)
rng2 = np.random.default_rng(42)

print(rng1.integers(1, 100, size=5))
print(rng2.integers(1, 100, size=5))
```

Important concept
Same seed + same operations = same random sequence.

Interview question: What is reproducibility in data analysis?

Answer: Reproducibility means getting the same results again when we use the same data, seed and operations.

```python
#rng.choice() – Random sampling
players = np.array([
    "Virat",
    "Rohit",
    "Gill",
    "Hardik",
    "Jadeja",
    "Bumrah"
])
```

```python
#Randomly 3 players select karna:
```

```python
rng = np.random.default_rng(42)

sample = rng.choice(players, size=3)

print(sample)
```

5. replace=True vs replace=False
important interview concept hai.

Example A – Without replacement

```python
rng = np.random.default_rng(42)

sample = rng.choice(
    players,
    size=3,
    replace=False
)

print(sample)
#Same player repeat nahi hoga.



#Har selected item unique hoga.
```

B – With replacement

```python
rng = np.random.default_rng(42)

sample = rng.choice(
    players,
    size=5,
    replace=True
)

print(sample)
```

```python
"""- Same player multiple times aa sakta hai.
- Selection ke baad item population mein available rehta hai.
Parameter	Meaning
replace=True	Repeated selection allowed
replace=False	Repeated selection not allowed"""
```

```python
#Actual IPL dataset practical
#dono uploaded datasets mein se delivery records
#  randomly select karenge.
```

```python
import numpy as np
import pandas as pd

matches = pd.read_csv("ipl_matches.csv")
deliveries = pd.read_csv("ipl_deliveries.csv")

# Random generator
rng = np.random.default_rng(26)

# Select 10 unique row positions
sample_indices = rng.choice(
    len(deliveries),
    size=10,
    replace=False
)

# Extract sampled records
audit_sample = deliveries.iloc[sample_indices]

print(audit_sample)
""""default_rng(26) – Reproducible random generator.



len(deliveries) – Total available delivery records.



size=10 – Ten records select karna.



replace=False – Duplicate selection avoid karna.



iloc – Selected row positions retrieve karna."""
```

7. Match dataset se random sampling

```python
rng = np.random.default_rng(26)

sample_indices = rng.choice(
    len(matches),
    size=5,
    replace=False
)

match_sample = matches.iloc[sample_indices]

print(match_sample[
    ["match_id", "team1", "team2", "winner"]
])  
#match-level sampling aur delivery-level sampling alag
```

Advanced practical – Sampled deliveries with match information

```python
audit_sample = deliveries.iloc[sample_indices]

audit_report = audit_sample.merge(
    matches[["match_id", "winner", "season"]],
    on="match_id",
    how="left"
)

print(audit_report)
```
