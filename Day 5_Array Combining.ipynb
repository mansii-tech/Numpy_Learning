Copy vs View – Interview Important
NumPy ka bahut important concept hai.

Copy: Original array ki independent copy create hoti hai. Copy mein changes karne se original array change nahi hota.

```python
arr = np.array([10, 20, 30])

new_arr = arr.copy()

new_arr[0] = 100

print("Original:", arr)
print("Copy:", new_arr)
```

```python
#View: Original array ka data share karne wala naya array object
#  create hota hai. View mein changes karne se original array bhi 
# change ho sakta hai.
```

```python
arr = np.array([10, 20, 30])

new_arr = arr.view()

new_arr[0] = 100

print("Original:", arr)
print("View:", new_arr)
```

Copy vs View diagram
COPY

Original Array
\(10, 20, 30\)

Independent copy

Copy Array
\(100, 20, 30\)

(Original unchanged)

VIEW

Original Array
\(100, 20, 30\)

Shared data

View Array
\(100, 20, 30\)

(Original can change)

Important technical point: Basic slicing generally creates a view, while advanced indexing generally creates a copy. reshape() may return either a view or a copy depending on memory layout.

Industry Practical – Combine Monthly Sales
Company ke paas January aur February ki sales hain. Humein dono datasets combine karne hain.


```python
import numpy as np

january = np.array([10000, 15000, 20000])
february = np.array([12000, 18000, 25000])

# Combine sales
all_sales = np.concatenate((january, february))

print("Combined Sales:", all_sales)

# Total sales
print("Total Sales:", np.sum(all_sales))

# Average sales
print("Average Sales:", np.mean(all_sales))
```

```python
#Industry applications:
#- Combining datasets from different months.
#- Merging batches of numerical data before analysis.
#- Splitting datasets into smaller parts.
#- Creating independent copies before modifying data.
```
