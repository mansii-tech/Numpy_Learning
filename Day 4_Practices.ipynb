```python
#reshape--kya--hota--hai#array ke elemnts
#  ko change kiye bina uska structure ya shape change karna 
```

```python
arr=np.array([1,2,3,4,5,6])
new_arr = arr.reshape(2,3)
new_arr
#1 d array ko 2 rows aur 3 columns me convert kiya
```

```python
arr = np.arange(1,13)
arr.reshape(3,4)
arr.reshape(4,3)
arr.reshape(2,2,3)
#reshape karte time total elment same hone chahiye
#12 element-- reshape (3,4) valid
#12 elment--reshape(2,6)valid
#12 elment --reshape(3,5)invalid
```

```python
#using 1 shape
#numpy automatically dimention calculate kar sakta hai
arr = np.arange(1,13)
result = arr.reshape(3,-1)
result
# -1 meaning numpy remaining dimensions khud calculate karega
```

```python
#flatten in numpy
"""2D ya multidimensional array ko 
single 1D array me canvert karta hai"""
arr = np.array([
    [1,2,3],
    [4,5,6]
])
result = arr.flatten()
print(result)
```

Reshape
#array ka structure change karta hai
#multiple shapes possible hai
#elments same rahte hai

Flatten
#multidimential array ko 1 D array banata hai
# output always one D array me hota hai
#elements same hote hai
Note--flatten()ek copy return karta hai
original array me changes nahi hote agar flatten array modify kare



```python
#Transpose in numpy
"""Rows ko column aur columns ko rows me 
convert karta hai"""
arr = np.array([[1,2,3],
               [4,5,6]])
result = arr.T
print(result)
print(arr.shape)
print(result.shape)
```

```python
#students mark data


marks = np.array([
    [80, 85, 90],
    [70, 75, 80],
    [90, 95, 85]
])

print("Original:")
print(marks)

print("Transpose:")
print(marks.T)
```

```python
#Practical Industry Case Study
"""Problem: Company ke paas 3 employees 
aur 3 months ki sales hain.
Humein har employee ki monthly sales mein
2,000 add karna hai."""
```

```python
import numpy as np

sales = np.array([
    [10000, 12000, 15000],
    [20000, 22000, 25000],
    [15000, 18000, 20000]
])

# Broadcasting
updated_sales = sales + 2000

# Shape
print("Shape:", updated_sales.shape)

# Transpose
print("Transpose:")
print(updated_sales.T)

# Flatten
print("Flatten:")
print(updated_sales.flatten())
```
