```python
#Part 3: Axis Operations – Most important
#Suppose 3 branches ki 3 months ki sales:
sales = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [200, 300, 400]
])
```

Sales table
Branch	January	February	March
A	    100	    200     	300
B	    150	    250	        350
C	    200	    300	        400





```python
#axis=0 – Column-wise calculation
print(np.sum(sales, axis=0))
```

```python
Dry run:
- January = 100 + 150 + 200 = 450
- February = 200 + 250 + 300 = 750
- March = 300 + 350 + 400 = 1050
Interpretation: Har month ki total sales.
```

```python
12. axis=1 – Row-wise calculation
print(np.sum(sales, axis=1))
```

Dry run:
- Branch A = 100 + 200 + 300 = 600
- Branch B = 150 + 250 + 350 = 750
- Branch C = 200 + 300 + 400 = 900
Interpretation: Har branch ki total sales.
Axis ka golden rule
Code	Calculation	Output
axis=0	Column-wise	[450, 750, 1050]
axis=1	Row-wise	[600, 750, 900]

Yaad rakho:
- axis=0: rows collapse hoti hain, columns remain karte hain.
- axis=1: columns collapse hote hain, rows remain karti hain.

Part 5: Real-world mini project – Retail Sales Analytics
Scenario: A company has 4 branches and 5 months of sales. Management wants to identify the best-performing branch and best-performing month.

```python
import numpy as np

sales = np.array([
    [12000, 15000, 18000, 14000, 20000],
    [10000, 17000, 16000, 19000, 21000],
    [14000, 13000, 20000, 22000, 25000],
    [18000, 16000, 19000, 24000, 28000]
])

# 1. Total sales of each branch
branch_total = np.sum(sales, axis=1)

# 2. Total sales of each month
monthly_total = np.sum(sales, axis=0)

# 3. Average sales of each branch
branch_avg = np.mean(sales, axis=1)

# 4. Highest and lowest individual sales
highest = np.max(sales)
lowest = np.min(sales)

# 5. Best performing branch index
best_branch = np.argmax(branch_total)

# 6. Best performing month index
best_month = np.argmax(monthly_total)

# 7. Sort branch totals
sorted_branches = np.sort(branch_total)[::-1]

print("Branch Total:", branch_total)
print("Monthly Total:", monthly_total)
print("Branch Average:", branch_avg)
print("Highest Sale:", highest)
print("Lowest Sale:", lowest)
print("Best Branch Index:", best_branch)
print("Best Month Index:", best_month)
print("Sorted Branch Sales:", sorted_branches)
```

1. Kaunsi branch ki total sales highest hai?
2. Kaunsa month best perform kar raha hai?
3. Kya highest individual sale wali branch hi overall best branch hai?
4. Average sales ke basis par branches ko rank karo.
5. Kya sales consistent hain ya fluctuate kar rahi hain?
