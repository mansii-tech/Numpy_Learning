5 employee ki salary me 10% increament karna hai

```python
salary =[20000,25000,30000,35000,40000]#WITH LOOP IN PYTHON
new_salary =[]
for s in salary:
    new_salary.append(s*1.10)
print(new_salary)
```

```python
import numpy as np
```

```python
#numpy vectorisation
salary=np.array([20000,25000,30000,35000,40000])
new_salary=salary*1.10
print(new_salary)
```

```python
#salary increment
salary = np.array([20000,25000,30000,35000,40000,50000])
increment_salary=salary+5000
print(increment_salary)
```

```python
#product discount
prices = np.array([1000,2000,3000,4000])
discounted_prices = prices *0.90
print(discounted_prices)#10% disscount apply
```
