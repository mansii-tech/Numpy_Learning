Array properties

```python
arr=np.array([10,20,30,40,50])
print("Shape",arr.shape)
print("Dimention",arr.ndim)
print("Size",arr.size)
print("Data type",arr.dtype)
```

```python
#2d array
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr.shape)
print(arr.ndim)
print(arr.size)
```
